```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from sentence_transformers import SentenceTransformer, util

# ۱. مدل تولید متن فارسی
model_name = "ai-forever/mGPT-1.3B-persian"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
generator = pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=50)

# ۲. مدل Sentence-BERT فارسی برای یافتن جمله مرتبط
embedder = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# ۳. گرفتن متن زمینه
print("متن زمینه را وارد کنید:")
context = input()

# شکستن متن به جملات
sentences = [s.strip() for s in context.split('.') if s.strip()]

print("\nمی‌توانید سوال‌های خود را بپرسید. برای خروج 'e' یا 'exit' را تایپ کنید.\n")

while True:
    question = input("سوال: ")
    if question.lower() in ["e", "exit", "خروج"]:
        print("خروج از برنامه. خداحافظ!")
        break

    # ۴. پیدا کردن جمله مرتبط با سوال
    question_emb = embedder.encode(question, convert_to_tensor=True)
    sentence_embs = embedder.encode(sentences, convert_to_tensor=True)
    cosine_scores = util.pytorch_cos_sim(question_emb, sentence_embs)[0]
    best_idx = cosine_scores.argmax().item()
    relevant_sentence = sentences[best_idx]

    # ۵. ساخت prompt برای مدل generative
    prompt = f"""
متن: {relevant_sentence}
سوال: {question}
پاسخ:
"""

    # ۶. تولید پاسخ
    response = generator(prompt, do_sample=False, max_new_tokens=50)[0]['generated_text']
    first_line = response.strip().split('\n')[0]
    print(first_line, "\n")
```
